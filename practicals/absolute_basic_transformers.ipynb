# Toy Example 1: Text Generation (Generative LLM)

- This example uses GPT-2, a lightweight generative transformer model that auto-completes text given a prompt.

```python
from transformers import pipeline

# 1. Load the text-generation pipeline with GPT-2
generator = pipeline("text-generation", model="gpt2")

# 2. Define a simple prompt
prompt = "In the future, artificial intelligence will"

# 3. Generate text using the transformer
output = generator(prompt, max_length=30, num_return_sequences=1)

# 4. Display the result
print(output[0]['generated_text'])
```

# Explanation

### The Pipeline Concept
Think of `pipeline()` as an all-in-one box. You put text in, and it handles everything required to return an answer.

### Behind the Scenes (3 Steps)
1. **Tokenizer**: Converts human words into numerical IDs (tokens) that the machine understands.
2. **Transformer Model**: Processes these tokens through self-attention layers to predict probabilities.
3. **Decoder**: Converts predicted numerical IDs back into readable words or label categories.

# Example 2: Sentiment Classification

```python
from transformers import pipeline

# 1. Load a pre-trained classification model
classifier = pipeline("sentiment-analysis")

# 2. Test positive and negative inputs
sample_text_1 = "Learning Python and machine learning is super fun!"
sample_text_2 = "I had a terrible experience with the slow network."

print(classifier(sample_text_1))
print(classifier(sample_text_2))
```

# Visualize the core mechanism of transformers—self-attention

- Parallel Processing: Instead of reading words one-by-one (like older models), it takes the whole sentence at once.

- Contextual Awareness: It shows how the word "CAT" computes its relationship to every other word to understand that it is sitting on a mat and is comfortable.

- Result: The final "Contextual Embeddings" demonstrate that the output for the word "CAT" isn't just a dictionary definition, but an embedding that has "absorbed" relevant context from the rest of the sentence.
